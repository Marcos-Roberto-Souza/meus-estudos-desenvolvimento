## Pandas

```python
import pandas as pd
```

```python
cidades = [
    "São Paulo (SP)",
    "Rio de Janeiro (RJ)",
    "Brasília (DF)",
    "Fortaleza (CE)",
    "Salvador (BA)",
    "Belo Horizonte (MG)",
    "Manaus (AM)",
    "Curitiba (PR)",    
]

populacoes = [
    12,
    6.5,
    3,
    2.4,
    2.4,
    2.3,
    2,
    1.7    
]
```

```python
cidades_serie = pd.Series(cidades)
cidades_serie
```

```python
populacoes_serie = pd.Series(populacoes)
populacoes_serie
```

```python
cidades_serie[3]
```

```python
cidades_serie[3:6]
```

## DataFrames

```python
cidades_populosas = pd.DataFrame({'cidade': cidades_serie, 'populacao em milhoes': populacoes_serie})
cidades_populosas
```

```python
area_urbana = [
    '914',
    '640',
    '590',
    '253',
    '196',
    '274',
    '277',
    '336'
]
```

```python
cidades_populosas['area urbana em km²'] = area_urbana
cidades_populosas
```

## Aplicando Funções

```python
cidades_populosas['populaca'] = cidades_populosas['populacao em milhoes'] * 1_000_000
cidades_populosas
```

```python
cidades_populosas.info()
```

```python
cidades_populosas['area urbana em km²'] = cidades_populosas['area urbana em km²'].astype(int)
cidades_populosas.info()
```

## Combinando Dados de Colunas

```python
cidades_populosas['habitantes por km²'] = cidades_populosas['populaca'] / cidades_populosas['area urbana em km²']
cidades_populosas
```
