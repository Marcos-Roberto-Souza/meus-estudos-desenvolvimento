# Lidando com Dados Faltantes

##### Imagine fazer uma análise para prever o preço de casas, faltando o número de quartos em alguns exemplos de dados. A análise levaria a resultados imprecisos e consequentemente a decisões erradas. Neste notebook veremos como lidar com dados faltantes em um conjunto de dados da área da saúde. 

```python
import pandas as pd
```

### Carregando os Dados

```python
diabetes = pd.read_excel('diabetes.xlsx')
```

```python
diabetes.to_csv('diabetes.csv', index=False)
```

```python
diabetes['diabetico'].value_counts()
```

```python
diabetes[diabetes['concentraçao de glicose'].isna() | diabetes['imc'].isna()]
```

```python
diabetes.dropna(axis='index', subset=['concentraçao de glicose', 'imc'], inplace=True)
```

```python
diabetes.info()
```

## Substituindo Valores

```python
diabetes.isna().sum()
```

```python
diabetes[diabetes['insulina'].isna()].head(20)
```

```python
diabetes['leu insulina'] = ~diabetes ['insulina'].isna()
diabetes
```

```python
diabetes.drop(columns=['insulina'], inplace=True)
diabetes
```

```python
diabetes = diabetes[[
    'pressao arterial',
    'concentraçao de glicose',
    'leu insulina',
    'tipo sanguineo',
    'idade',
    'imc',
    'diabetico'
]]
diabetes
```

```python
diabetes.info()
```

## Preenchendo Valores

```python
diabetes.isna().sum()
```

```python
diabetes['pressao arterial']
```
