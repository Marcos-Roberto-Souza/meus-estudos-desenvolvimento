```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dados = pd.read_excel("sentiment_dataset.xlsx")
```

```python
dados.to_csv("sentiment_dataset.csv", index=False)
```

```python
df = pd.read_csv("sentiment_dataset.csv")
df.head()
```

```python
df["Country"].value_counts()
```

```python
df["Sentiment"].values
```

```python
df["Country"].values
```

```python
df["Sentiment"] = df["Sentiment"].str.strip()
df["Country"] = df["Country"].str.strip()
```

```python
df_new = df[df["Country"] == 'USA']
```

```python
fig, ax = plt.subplots()
ax.boxplot([df_new["Likes"]], labels=["USA"])
ax.set_title("País X Likes")
plt.show()
```

## Filtra as linhas com sentimento 'positivo' do país 'USA'

```python
df_new3 = df_new[df_new["Sentiment"] == "Positive"]
```

```python
df_new["Sentiment"].value_counts()
```

### Subplots de boxplots

```python
fig, ax = plt.subplots()
ax.boxplot([df_new["Likes"], df_new3["Likes"]], labels=["USA", "Positive"])
ax.set_title("País x Likes")
plt.show()
```
