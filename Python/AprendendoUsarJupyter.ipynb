```python
# multiplicação de matrizes
A = np.array([[1, 2],[3, 4], [5, 6]])
B = np.array([[1], [2]])

'''
print(f'lista de A:\n{A}')

print(f'lista de B:\n{B}')
'''

```

```python
## Multiplicação utilizando apenas listas python
C = [[0], [0], [0]]

for i in range(len(A)):
    for j in range(len(B[0])):
        for k in range(len(B)):
            C[i][j]+= A[i][k] * B[k][j]

print(C)
```

```python
## agora vamos realizar a mesma operação com NumPy
D = A @ B
print(D)
```

```python
N = 500

A = np.arange(N*N, dtype='int64').reshape(N, N)
B = A * 2

C = A @ B

print(C)
## Tempo de execucao: 0.196
## Memoria utilizada: 5.8 Mb
```

```python
A_lst = A.tolist()
B_lst = B.tolist()
C_lst = np.zeros_like(A).tolist()

for i in range(len(A_lst)):
    for j in range(len(B_lst[0])):
        for k in range(len(B_lst)):
            C_lst[i][j] += A_lst[i][k] * B_lst[k][j]


## tempo de execução: 66.462s
## Memoria utilizada: 211.6 Mb
```
