### Explorando a Diferença Prática entre Média, Mediana e Moda

```python
lst = [2, 4, 6,1, 3, 0, 4, 3, 5, 3, 5, 4]
sum(lst) / len(lst)
```

```python
print(len(lst))
```

```python
import statistics

statistics.mean(lst)
```

```python
lst_sorted = sorted(lst)
print(lst_sorted)
```

```python
s = len(lst_sorted)
m = s//2

if s % 2!=0:
    print(lst_sorted[m])
else:
    print((lst_sorted[m - 1] + lst_sorted[m])/2)
```

```python
#mediana tamanho par
statistics.median(lst)
```

```python
mediana_impar = lst + [10]
mediana_impar
```

```python
statistics.median(mediana_impar)
```

```python
freq = {}
for n in lst:
    if n in freq:
        freq[n] += 1
    else:
        freq[n] = 1

freq    
```

```python
max(freq, key=freq.get)
```

```python
statistics.mode(lst)
```
