# Importando as bibliotecas e Lendo o dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("meteorite_landings.csv")
df.head()
```

## Histograma da latitude

```python
df_new = df[df["reclong"] < 300] # removendo outliers
```

```python
plt.hist(df_new["reclat"])
plt.show()
```

### Variando a Cor

```python
plt.hist(df_new["reclong"], color="red")
plt.show()
```

## Histogramas em subplots

```python
plt.figure(figsize=(15,5))
plt.subplot(1,2,1) #linha, coluna, contagem
plt.hist(df_new["reclong"])
plt.xlabel("Longitude")
plt.ylabel("Frequência")
plt.subplot(1,2,2) #Linha, coluna, contagem
plt.hist(df_new["reclat"], color="green")
plt.xlabel("Latitude")
plt.ylabel("Frequência")
plt.show()
```

### latitude e Longitude no mesmo gráfico

```python
plt.figure(figsize=(15,5))
plt.hist([df['reclat'], df['reclong']], bins=20, label=['lat', 'long'])
plt.legend(loc='upper right')
plt.show()
```
