# Exemplo if

```python
x = 5
y = 6

if x < y: # aqui entra a condição que if(se) x for menor(<) que y - mostra o x(print)
    print(x)
if y > x: # aqui entra a condição que if(se) y for maior(<) que x - mostra o y(print)
    print(y)
if x > y: # aqui entra a condição falsa mas o programa não imprime na tela porque é uma condição falsa não da erro mas também não imprime por que x é menor.
    print(x)
```

# Exemplo if e else

```python
primeiroNumero = int(input())

if primeiroNumero%2==0:# aqui ele verifica se o numero digitado é par
    print("E par")
else:# aqui a condição cai se o numero for impar
    print("E impar")
```

```python
nomeAnimal = input()

if nomeAnimal == 'macaco': # aqui ele verifica se o que digitou esta na variável 
    print('símio ou mono são designações genéricas e ambíguas para vários mamíferos da infraordem dos simiiformes')
else: #aqui ele imprime se a condição decima não for verdadeira
    print("Animal não existe no banco de Dados")
```

# Exemplo if, elif e else

```python
numeroDigitado = float(input())

if numeroDigitado >= 100 and numeroDigitado < 1300: # aqui a condição verifica se o numero esta entre 100 a 1299, se não for vai para próxima condição
    print("Você Faz Bico")
elif numeroDigitado == 1300: # aqui a condição verifica se o numero é igual a 1300, se não for verdadeira a condição vai para próxima condição
    print("Você Ganha Salário Minímo")
elif numeroDigitado > 1300 and numeroDigitado <= 2500: # aqui a condição verifica se o numero está de 1301 á 2500, se não vai para próxima condição
    print("Você Ganha um Salário e Meio")
elif numeroDigitado > 2500 and numeroDigitado <= 4000: # aqui a condição verifica se o numero está de 2501 á 4000, se não vai para próxima condição
    print("Você Ganha 2 a 3 salarios Minímos")
elif numeroDigitado > 4000: # aqui a condição verifica se o número está acima de 4001, se não vai para próxima condição
    print("Você é assalariado ganha muito Dinheiro")
else: # aqui é a condição que nenhuma das outras foi verdadeira
    print("Você é Pobre ou esta Desempregado")
```
